# Create Domain Patterns

Quick visual check of `gs_generator` outputs. Requires `numpy`, `matplotlib` (and optionally `cupy` for GPU).

```python
import matplotlib.pyplot as plt
import numpy as np



import sys
from pathlib import Path

repo_root = Path.cwd()

# Also works when Jupyter starts inside tutorials/
if not (repo_root / "src").exists():
    repo_root = repo_root.parent

src_dir = repo_root / "src"
if not src_dir.exists():
    raise RuntimeError(f"Cannot find repository src directory from {Path.cwd()}")

if str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))
    
from scattering_calculator.sample_generator.gray_scott_generator import generate
from scattering_calculator.sample_generator.gray_scott_generator_binary import generate as generate_binary
plt.rcParams["figure.constrained_layout.use"] = True
```

## 1. Basic labyrinth batch

```python
A, B, meta = generate(
    batch=16,
    H=100,
    W=100,
    n_steps=1000,#6000
    region="labyrinth",
    use_gpu=False,  # flip to False if CuPy is not installed
    seed=42,
)

fig, axes = plt.subplots(4, 4, figsize=(9, 9))
for i, ax in enumerate(axes.flat):
    ax.imshow(B[i], cmap="gray")
    ax.set_title(f"f={meta['f'][i]:.4f}  k={meta['k'][i]:.4f}", fontsize=9)
    ax.axis("off")
plt.suptitle("Gray-Scott labyrinths")
```

## 2. Diverse transient states

```python
A, B, meta = generate(
    batch=8,
    H=192,
    W=192,
    n_steps=6000,
    region="labyrinth",
    random_stop=True,
    random_stop_min_frac=0.15,
    use_gpu=False,
    seed=11,
)

fig, axes = plt.subplots(2, 4, figsize=(14, 7))
for i, ax in enumerate(axes.flat):
    ax.imshow(B[i], cmap="gray")
    ax.set_title(f"stop={int(meta['stop_steps'][i])}", fontsize=9)
    ax.axis("off")
plt.suptitle("Random-stop transient states")
plt.show()

print("Saved: check_labyrinth.png, check_random_stop.png")
```
